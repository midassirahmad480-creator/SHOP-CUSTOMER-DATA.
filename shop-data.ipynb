```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/kaggle/input/datasets/mudasir123312/shop-customer-data/Customers.csv")
```

```python
print("Rows and Columns:", df.shape)
```

```python
print(df.isnull().sum())
```

```python
df.describe()
```

```python
print("Average Age:", df['Age'].mean())
```

```python
print("Average Annual Income:",
      df['Annual Income ($)'].mean())
```

```python
print("Average Spending Score:",
      df['Spending Score (1-100)'].mean())
```

```python
df['Gender'].value_counts()
```

```python
sns.countplot(data=df, x='Gender')

plt.title('Customers by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')

plt.show()
```

```python
df['Profession'].value_counts()
```

```python
plt.figure(figsize=(10,6))

sns.countplot(data=df, y='Profession')

plt.title('Customers by Profession')
plt.xlabel('Number of Customers')
plt.ylabel('Profession')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Age'], bins=20)

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Customers')

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='Annual Income ($)',
    y='Spending Score (1-100)',
    hue='Gender'
)

plt.title('Annual Income vs Spending Score')

plt.show()
```

```python
plt.figure(figsize=(9,6))

numeric_df = df.select_dtypes(include='number')

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()
```

```python
top_customers = df.sort_values(
    by='Spending Score (1-100)',
    ascending=False
).head(10)

display(top_customers)
```
